```python
models = [optimised_DT, adaboost_classifier, best_gradientboosting_classifier]
model_names = ['random_forest', 'AdaBoost', 'GBoost']
fnr_values = []

for model in models:
    y_pred = model.predict(test_x)
    # early stage cancer is the positive class, so ravel() returns tn, fp, fn, tp
    tn, fp, fn, tp = confusion_matrix(test_y, y_pred, labels=['healthy', 'early stage cancer']).ravel()
    fnr_values.append(fn / (fn + tp)) # share of early-stage cancers predicted as healthy

plt.bar(model_names, fnr_values)
plt.ylabel('False Negative Rate')
plt.title('FNR Comparison')
plt.show()
```
