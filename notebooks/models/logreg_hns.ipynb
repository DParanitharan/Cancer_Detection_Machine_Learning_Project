A Pipeline is built using RandomUnderSampler and SMOTE from the 'imblearn' package to randomly undersample the majority class and the create synthetuc data for the minority class simultaneously. Logistic Regression model is initialised.

```python
from imblearn.over_sampling import SMOTE
over = SMOTE(sampling_strategy = 0.7, random_state=42)
under = RandomUnderSampler(sampling_strategy = 0.98, random_state=42)
from imblearn.pipeline import Pipeline
logreg_model = make_pipeline(StandardScaler(), LogisticRegression(solver='liblinear', random_state=42)) # scaler is fitted on training data only
steps = [('o', over), ('u', under)]
pipeline = Pipeline(steps=steps)
X_over, y_over = pipeline.fit_resample(xvals_train_filtered, yvals_train_binary)
logreg_model.fit(X_over, y_over)
logreg_model[-1].coef_, logreg_model[-1].intercept_
```

```python
testX_over, testY_over = xvals_test_filtered, yvals_test_binary  # evaluate on the real (imbalanced) test set
```

The summary of the logistic regression model is as shown below

```python
print(confusion_matrix(yvals_test_binary, logreg_model.predict(xvals_test_filtered)))
print(classification_report(yvals_test_binary, logreg_model.predict(xvals_test_filtered)))
```

Precision-recall curve is plotted to evaluate performance of model.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import f1_score
from sklearn.metrics import auc
# predict probabilities
lr_probs = logreg_model.predict_proba(testX_over)
# keep probabilities for the positive outcome only
lr_probs = lr_probs[:, 1]
# predict class values
yhat = logreg_model.predict(testX_over)
lr_precision, lr_recall, _ = precision_recall_curve(testY_over, lr_probs)
lr_f1, lr_auc = f1_score(testY_over, yhat), auc(lr_recall, lr_precision)
# summarize scores
print('Logistic: f1=%.3f auc=%.3f' % (lr_f1, lr_auc))
# plot the precision-recall curves
no_skill = len(testY_over[testY_over==1]) / len(testY_over)
plt.plot([0, 1], [no_skill, no_skill], linestyle='--', label='No Skill')
plt.plot(lr_recall, lr_precision, marker='.', label='Logistic')
# axis labels
plt.xlabel('Recall')
plt.ylabel('Precision')
# show the legend
plt.legend()
# show the plot
plt.show()
```
